### Detailed Stats on Normal Distribution

```python
np_normal_dis = np.random.normal(5, 0.5, 1000)
print('min:', np.min(np_normal_dis))
print('max:', np.max(np_normal_dis))
print('mean:', np.mean(np_normal_dis))
print('median:', np.median(np_normal_dis))
# Mode for continuous data may not be meaningful, but we compute it
mode_result = mode(np_normal_dis, keepdims=True)  # Updated for newer SciPy
print('mode:', mode_result)
print('std:', np.std(np_normal_dis))
```

```python
plt.figure(figsize=(8, 4))
plt.hist(np_normal_dis, color="grey", bins=21)
plt.title("Normal Distribution (mu=5, sigma=0.5)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

### Linear Algebra

```python
f = np.array([1, 2, 3])
g = np.array([4, 5, 3])
dot_product = np.dot(f, g)
print('Dot product of [1,2,3] and [4,5,3]:', dot_product)
```

### Checkerboard Zeros

```python
Z = np.zeros((8, 8))
Z[1::2, ::2] = 1  # Odd rows, even cols
Z[::2, 1::2] = 1  # Even rows, odd cols
print('Checkerboard pattern:\n', Z)
```

### Linear Relationship Example

```python
temp = np.array([1, 2, 3, 4, 5])
pressure = temp * 2 + 5
print('Pressure:', pressure)

plt.figure(figsize=(8, 4))
plt.plot(temp, pressure)
plt.xlabel('Temperature in °C')
plt.ylabel('Pressure in atm')
plt.title('Temperature vs Pressure')
plt.xticks(np.arange(0, 6, step=0.5))
plt.grid(True)
plt.show()
```

### Gaussian Distribution Plot

```python
mu = 28
sigma = 15
samples = 100000
x = np.random.normal(mu, sigma, samples)

plt.figure(figsize=(9, 5))
ax = sns.distplot(x, bins=100, color='blue')
ax.set(xlabel="x", ylabel="Density", title="Gaussian Normal Distribution (mu=28, sigma=15)")
plt.show()
```
